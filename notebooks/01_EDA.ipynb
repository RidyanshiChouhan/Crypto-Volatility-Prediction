# 01 - Exploratory Data Analysis

Dataset overview, statistics, and Plotly visualizations.

```python
import sys
from pathlib import Path
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))

from src.preprocessing import load_processed
from src.config import REPORTS_ASSETS_DIR

df = load_processed('processed_combined.csv')
REPORTS_ASSETS_DIR.mkdir(parents=True, exist_ok=True)
df.head()
```

## Dataset Overview

```python
print('Shape:', df.shape)
print(df.dtypes)
print('Missing:', df.isnull().sum().sort_values(ascending=False).head(10))
```

## Statistical Summary

```python
df.describe().T
```

## Visualizations

```python
fig = px.line(df, x='date', y='close', color='symbol', title='Price Trends')
fig.write_image(str(REPORTS_ASSETS_DIR / 'price_trends.png'))
fig.show()

fig = px.histogram(df, x='rolling_volatility_7', color='symbol', nbins=40, title='Volatility Distribution')
fig.write_image(str(REPORTS_ASSETS_DIR / 'volatility_distribution.png'))
fig.show()
```
